# 02 – Data Quality Analysis

This notebook evaluates missing values, duplicate records, data types and dataset dimensions in the cleaned Olist datasets.

```python

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/content/Cart2Insights")
CLEAN = ROOT / "data" / "cleaned"
RAW = ROOT / "data" / "raw"

if not CLEAN.exists():
    raise FileNotFoundError("Project data/cleaned folder was not found.")

print("Cleaned data folder:", CLEAN)
```

```python

datasets = {
    "customers": "customers_cleaned.csv",
    "products": "products_cleaned.csv",
    "orders": "orders_cleaned.csv",
    "payments": "payments_cleaned.csv",
    "reviews": "reviews_cleaned.csv",
    "sellers": "sellers_cleaned.csv",
    "order_items": "order_items_cleaned.csv",
    "category_translation": "category_translation_cleaned.csv"
}

quality = []
for name, filename in datasets.items():
    df = pd.read_csv(CLEAN / filename, low_memory=False)
    quality.append({
        "dataset": name,
        "rows": len(df),
        "columns": len(df.columns),
        "missing_cells": int(df.isna().sum().sum()),
        "duplicate_rows": int(df.duplicated().sum())
    })

quality_report = pd.DataFrame(quality)
display(quality_report)
```

```python

for name, filename in datasets.items():
    df = pd.read_csv(CLEAN / filename, low_memory=False)
    print("\\n", name, "— data types")
    display(df.dtypes.astype(str).to_frame("data_type"))
```
