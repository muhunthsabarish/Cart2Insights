# 03 – Data Cleaning and Validation

The original raw datasets are preserved in `data/raw/`. The cleaned CSV datasets were generated separately and stored in `data/cleaned/`. This notebook validates the cleaned outputs without modifying the raw files.

```python

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/content/Cart2Insights")
CLEAN = ROOT / "data" / "cleaned"
RAW = ROOT / "data" / "raw"

if not CLEAN.exists():
    raise FileNotFoundError("Project data/cleaned folder was not found.")

print("Cleaned data folder:", CLEAN)
```

```python

expected_files = [
    "customers_cleaned.csv", "products_cleaned.csv",
    "orders_cleaned.csv", "payments_cleaned.csv",
    "reviews_cleaned.csv", "sellers_cleaned.csv",
    "order_items_cleaned.csv", "category_translation_cleaned.csv"
]

for filename in expected_files:
    path = CLEAN / filename
    if path.exists():
        df = pd.read_csv(path, low_memory=False)
        print(f"{filename}: {df.shape} — available")
    else:
        print(f"{filename}: MISSING")
```

```python

# Validate key fields and duplicate identifiers
checks = {
    "customers_cleaned.csv": "customer_id",
    "products_cleaned.csv": "product_id",
    "orders_cleaned.csv": "order_id",
    "sellers_cleaned.csv": "seller_id",
    "reviews_cleaned.csv": "review_id"
}

for filename, key in checks.items():
    df = pd.read_csv(CLEAN / filename, low_memory=False)
    print(filename, "| missing keys:", df[key].isna().sum(),
          "| duplicate keys:", df[key].duplicated().sum())
```

## Cleaning decisions

- Original raw files are retained separately.
- Duplicate records were checked and removed using relevant identifiers where applicable.
- Date and numeric fields were converted to appropriate types during cleaning.
- Missing product measurements were handled using median values.
- Invalid geolocation coordinates were excluded from the cleaned geolocation output.
