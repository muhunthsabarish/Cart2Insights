# 06 – Exploratory Data Analysis

Explore sales, order status and customer purchasing patterns using summary statistics and visualizations.

```python

from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path("/content/Cart2Insights")
CLEAN = ROOT / "data" / "cleaned"
RAW = ROOT / "data" / "raw"

if not CLEAN.exists():
    raise FileNotFoundError("Project data/cleaned folder was not found.")

print("Cleaned data folder:", CLEAN)
```

```python

import matplotlib.pyplot as plt

orders = pd.read_csv(CLEAN / "orders_cleaned.csv")
items = pd.read_csv(CLEAN / "order_items_cleaned.csv")

print("Order status distribution:")
display(orders["order_status"].value_counts().to_frame("order_count"))

print("Sales summary:")
display(items[["price", "freight_value"]].describe())
```

```python

# Order status chart
orders["order_status"].value_counts().plot(
    kind="bar", figsize=(9, 4), title="Order Status Distribution"
)
plt.xlabel("Order status")
plt.ylabel("Number of orders")
plt.tight_layout()
plt.show()
```

```python

# Monthly item revenue trend
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"], errors="coerce"
)
monthly_orders = orders[["order_id", "order_purchase_timestamp"]].merge(
    items.groupby("order_id", as_index=False)["price"].sum(),
    on="order_id", how="inner"
)
monthly_orders["month"] = monthly_orders[
    "order_purchase_timestamp"
].dt.to_period("M").astype(str)

monthly_revenue = monthly_orders.groupby("month")["price"].sum()
monthly_revenue.plot(figsize=(11, 4), marker="o",
                     title="Monthly Item Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
